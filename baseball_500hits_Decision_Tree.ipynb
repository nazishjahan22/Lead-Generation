```python
import pandas as pd
```

Import pandas as pd, thne we read the csv file called ‘500hits.csv’

```python
df= pd.read_csv("/content/500hits.csv", encoding='latin1')
```

```python
df.head()
```

 Drop the columns ‘Players and ‘cs’

```python
df = df.drop(columns = ['PLAYER', 'CS'])
```

Select all rows and columns 0 through 12 (not including 13)  and assign it to X

```python
X = df.iloc[:, 0:13]
```

Select all rows : from column index 13, and assign it to y

```python
y = df.iloc[:,13]
```

Import train_test_split from sklearn.model_selection

```python
from sklearn.model_selection import train_test_split
```

Split the data into testing and train sets.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=17, test_size=0.2)
```

Check the snape or dimension of the data, .shape returns (rows, columns)

```python
X_train.shape
```

```python
X_test.shape
```

Import DecisionTreeClassifier

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier()
```

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.get_params()
```

Train the model on our data using the .fit()

```python
dtc.fit(X_train, y_train)
```

After traing we predict, i.e we evaluate our models performance using the test data.

```python
y_pred = dtc.predict(X_test)
```

Import the confusion_matrix function from scikit-learn.

It is used to evaluate classification models by comparing the predicted labels with the actual labels.

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

Return an array of numbers showing how important each feature was in buiding the decision tree “dtc”

```python
features = pd.DataFrame(dtc.feature_importances_, index = X.columns)
```

```python
features.head(15)
```

Higher values of ccp_alpha lead to more pruning in the decision tree.
This is part of cost-complexity pruning, which removes less important branches to help prevent overfitting.



ccp_alpha(Cost Complexity Pruning Alpha) is a parameter that controls how much the tree should be simplified.

```python
dtc2 = DecisionTreeClassifier(criterion='entropy', ccp_alpha= 0.04)
```

After adding the ccp_alpha, we train, test and evaluate again to see if theres improvement.

```python
dtc2.fit(X_train, y_train)
```

```python
y_pred2 = dtc2.predict(X_test)
```

```python
print(confusion_matrix(y_test, y_pred2))
```

```python
print(classification_report(y_test, y_pred2))
```
